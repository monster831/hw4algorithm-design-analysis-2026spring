## 问题 1

根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

### 问题 1 答案

**解答与步骤**

主定理的标准形式为：$T(n) = aT(n/b) + f(n)$，其中 $a \ge 1, b > 1$。通过比较 $f(n)$ 与 $n^{\log_b a}$ 的增长速度来求解。

1. **$T(n) = T(n/4) + 1$**
   * **参数：** $a = 1, b = 4, f(n) = 1$。
   * **比较：** $n^{\log_b a} = n^{\log_4 1} = n^0 = 1$。
   * **结论：** 此时 $f(n) = \Theta(n^{\log_b a})$，属于主定理的情况 2。
   * **结果：** $T(n) = \Theta(n^{\log_b a} \log n) = \mathbf{\Theta(\log n)}$。

2. **$T(n) = 3T(n/3) + n^2$**
   * **参数：** $a = 3, b = 3, f(n) = n^2$。
   * **比较：** $n^{\log_b a} = n^{\log_3 3} = n^1$。
   * **结论：** 明显 $f(n) = n^2$ 比 $n^1$ 增长快。检查正则条件：$a f(n/b) = 3(n/3)^2 = \frac{1}{3}n^2 \le cf(n)$（当 $c = 1/3$ 时成立）。属于主定理的情况 3。
   * **结果：** $T(n) = \Theta(f(n)) = \mathbf{\Theta(n^2)}$。

3. **$T(n) = T(2n/3) + 1$**
   * **参数：** $a = 1, b = 3/2, f(n) = 1$。
   * **比较：** $n^{\log_{3/2} 1} = n^0 = 1$。
   * **结论：** 类似于第 1 题，$f(n) = \Theta(n^{\log_b a})$，属于主定理的情况 2。
   * **结果：** $T(n) = \Theta(\log_{3/2} n) = \mathbf{\Theta(\log n)}$。

4. **$T(n) = 5T(n/4) + n$**
   * **参数：** $a = 5, b = 4, f(n) = n$。
   * **比较：** $n^{\log_b a} = n^{\log_4 5}$。因为 $5 > 4$，所以 $\log_4 5 > 1$。
   * **结论：** $n^{\log_4 5} \approx n^{1.16}$，比 $f(n) = n^1$ 增长快，属于主定理的情况 1。
   * **结果：** $T(n) = \mathbf{\Theta(n^{\log_4 5})}$。

5. **$T(n) = T(\sqrt{n}) + 1$**
   * **求解：** 该式不符合标准主定理形式，使用变量代换法。设 $n = 2^m$，则 $m = \log_2 n$。
   * **代入：** $T(2^m) = T(2^{m/2}) + 1$。令 $S(m) = T(2^m)$，得到 $S(m) = S(m/2) + 1$。
   * **主定理：** 对 $S(m)$ 使用主定理（情况 2），得到 $S(m) = \Theta(\log m)$。
   * **结果：** 因为 $m = \log_2 n$，回代得到 $T(n) = \mathbf{\Theta(\log \log n)}$。

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

### 问题 3 答案

**算法设计思路（暴力枚举法）**

最直观的方案是遍历所有可能的买入和卖出日期组合，寻找能够产生最大总收益的区间。
1. 初始化三个变量：`max_profit` 记录当前找到的最大收益（初始负无穷），`buy_day` 和 `sell_day` 记录最佳买卖日期。
2. 外层循环 `i` 从 0 到 n-1，代表买入起始日。初始化当前累加收益 `current_sum = 0`。
3. 内层循环 `j` 从 i 到 n-1，代表卖出日。不断将 `profits[j]` 累加到 `current_sum`。
4. 每次累加后比较并更新 `max_profit`。

 **时间复杂度：** $O(n^2)$。
   **操作次数推导：** 算法包含两层嵌套循环。外层循环变量 $i$ 从 $0$ 遍历到 $n-1$，共执行 $n$ 次。内层循环变量 $j$ 从 $i$ 遍历到 $n-1$。对于给定的 $i$，内层循环执行的次数为 $n - i$ 次。
   因此，内层循环中的基本操作（累加和比较操作，耗时 $O(1)$）的总执行次数 $S$ 可以表示为一个等差数列求和：
$$S = \sum_{i=0}^{n-1} (n - i) = n + (n-1) + (n-2) + \dots + 1 = \frac{n(n+1)}{2}$$
   展开后得到 $\frac{1}{2}n^2 + \frac{1}{2}n$。忽略低阶项和常数系数，最高阶项为 $n^2$，因此其严格渐进时间复杂度为 $O(n^2)$。
* **空间复杂度：** $O(1)$。算法只使用了有限的几个常数级额外变量（`max_profit`, `buy_day`, `current_sum` 等），没有开辟额外的数据结构，因此空间复杂度为 $O(1)$。

````python
def max_profit_brute_force(profits):
    n = len(profits)
    
    # 初始化最大收益为负无穷，确保任何实际收益都能更新它
    max_profit = float('-inf')
    buy_day = -1
    sell_day = -1
    
    # i 代表买入的日期索引
    for i in range(n):
        current_sum = 0
        # j 代表卖出的日期索引（可以与买入为同一天，但按题意我们要找最大连续和）
        for j in range(i, n):
            current_sum += profits[j]
            
            # 如果当前累加收益大于已知最大收益，则更新记录
            if current_sum > max_profit:
                max_profit = current_sum
                # 题目要求输出第几天（1-based 索引），所以索引需要 +1
                buy_day = i + 1
                sell_day = j + 1
                
    return max_profit, buy_day, sell_day

# ================= 题目中的测试示例 =================
profits_example = [3, 2, 1, -7, 5, 2, -1, 3, -1]
profit, buy, sell = max_profit_brute_force(profits_example)

print(f"profits={profits_example}")
print(f"第{buy}天买入，第{sell}天卖出，收益最大：{profit}")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

### 问题 4 答案（此题求助了大模型）

**算法设计思路（二分查找）**

要在 $O(\log n)$ 的时间复杂度下找到目标，我们需要使用二分查找法。
题目规定 $A[0] > A[1]$（下降趋势），$A[n-2] < A[n-1]$（上升趋势）。两端向内看都是“往下走”和“往上收”，那么在索引 1 到 n-2 之间必然存在一个谷底。
1. 设 `left = 1`，`right = n - 2`。
2. 每次取中间位置 `mid = (left + right) // 2`。
3. 如果 `A[mid] < A[mid-1]` 且 `A[mid] < A[mid+1]`，说明找到了局部最小，直接返回。
4. 如果 `A[mid] > A[mid+1]`，说明目前处于下降趋势，右侧必有谷底，更新 `left = mid + 1`。
5. 如果 `A[mid] > A[mid-1]`，说明处于上升趋势，左侧必有谷底，更新 `right = mid - 1`。

**复杂度分析**

* **时间复杂度：** $O(\log n)$。每次比较后搜索区间减半。

````python
def find_local_minimum(A):
    """
    在 O(log n) 时间复杂度下找到数组中的一个局部最小值并返回其数值。
    """
    n = len(A)
    
    # 数组长度至少为3才能满足首部下降、尾部上升的条件
    if n < 3:
        return None 
        
    # 定义二分查找的左右边界（剔除首尾肯定不是局部最小的元素）
    left = 1
    right = n - 2
    
    while left <= right:
        mid = left + (right - left) // 2
        
        # 满足局部最小条件：比左边小，且比右边小
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]
            
        # 如果当前元素比右边的元素大，说明处于下降趋势，右侧必有谷底
        if A[mid] > A[mid + 1]:
            left = mid + 1
        # 否则说明处于上升趋势（由于相邻不相等，这里暗含 A[mid] > A[mid-1]），左侧必有谷底
        else:
            right = mid - 1
            
    return None

# ================= 测试运行 =================
# 题目给定的测试用例
A = [3, 2, 5, 3, 3, 4, 5]
result = find_local_minimum(A)

print(f"输入数组: {A}")
# 注意：该数组有3和1两个局部最小，由于题目只要求找到"一个"并返回，
# 二分查找会根据趋势快速锁定其中一个。对于此数组，算法会稳定输出 1。
print(f"找到的一个局部最小值为: {result}")
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

### 问题 5 答案（此题求助了大模型）

**算法设计思路与详细复杂度分析**

**1. 基于排序算法求前 k 小的数**
* **算法设计思路：** 我们选择平均性能优异的**快速排序 (Quick Sort)**。其基本思想是分治法：在数组中选择一个基准元素（pivot），将数组分为小于、等于和大于基准的三部分；递归地对左右部分进行排序；最后合并。数组完全有序后，直接通过切片输出前 $k$ 个元素。
* **时间复杂度详细推导：**
    * **排序阶段：** 快速排序在每次划分时，如果能较为均匀地将数组一分为二，递归树的深度为 $\log_2 n$，每层划分需要遍历整个子数组，时间开销为 $O(n)$。因此，平均情况下的时间复杂度为 $O(n \log n)$。最坏情况（如数组已有序）下退化为 $O(n^2)$，但可以通过随机选择基准来极大概率避免。
    * **输出阶段：** 从长度为 $n$ 的有序数组中切片取出前 $k$ 个元素，时间耗费为 $O(k)$。
    * **总时间复杂度：** $O(n \log n) + O(k)$。由于通常 $k \le n$，$O(k)$ 会被更高阶的 $O(n \log n)$ 吸收，最终平均时间复杂度记为 $\mathbf{O(n \log n)}$。
* **空间复杂度：** $O(\log n)$，主要来自于快速排序平均情况下的递归调用栈开销。

**2. 基于最小堆（Min-Heap）求前 k 小的数**
* **算法设计思路：** 严格按照题目要求的“往空堆中不断插入元素”的方法。首先初始化一个空数组作为堆；然后遍历原数组的 $n$ 个元素，将它们逐个插入堆中，并在每次插入后通过上浮（shift-up）操作维护最小堆的性质；建堆完成后，执行 $k$ 次提取堆顶（Extract-min）操作，每次提取出的就是当前堆中的最小值，并在提取后通过下沉（shift-down）操作恢复堆性质。
* **时间复杂度详细推导：**
    * **建堆阶段：** 向包含 $i$ 个元素的堆中插入一个新元素，最坏情况需要沿着树高上浮，时间开销为 $O(\log i)$。依次插入 $n$ 个元素，总的时间复杂度为：
        $$\sum_{i=1}^{n} \log_2 i = \log_2 1 + \log_2 2 + \dots + \log_2 n = \log_2(n!)$$
        根据**斯特林近似公式（Stirling's approximation）**，$\log_2(n!) = \Theta(n \log n)$。因此，这种“逐个插入”的建堆方法时间复杂度为 $O(n \log n)$。（注：如果采用自底向上的直接 heapify 法，建堆仅需 $O(n)$，但题目限定了建堆方式）。
    * **提取阶段：** 每次调用 Extract-min 后，需要将堆尾元素移到堆顶并下沉，单次时间复杂度为 $O(\log n)$。调用 $k$ 次，总时间为 $O(k \log n)$。
    * **总时间复杂度：** 建堆 $O(n \log n)$ + 提取 $O(k \log n)$ = $O((n+k)\log n)$。同样，由于 $k \le n$，提取公因式后，整体时间复杂度简化为 $\mathbf{O(n \log n)}$。
* **空间复杂度：** $O(n)$，需要开辟一个大小为 $n$ 的数组来存储这棵完全二叉树（最小堆）。

1、基于快速排序求前 k 小的数

````python
def quick_sort(arr):
    """实现快速排序算法"""
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

def get_k_smallest_by_sorting(arr, k):
    """先排序，再输出前 k 个最小数"""
    if k <= 0 or not arr:
        return []
    sorted_arr = quick_sort(arr)
    return sorted_arr[:k]

# ================= 测试 =================
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
result = get_k_smallest_by_sorting(arr, k)

arr_str = "[" + ",".join(map(str, arr)) + "]"
result_str = ", ".join(map(str, result))
print(f"数组arr={arr_str} 的前{k}个最小数据为：{result_str}")
````

2、基于最小堆（Heap）求前 k 小的数

````python
import heapq

def get_k_smallest_by_heap(arr, k):
    """使用不断插入建堆的方法，再提取 k 次"""
    if k <= 0 or not arr:
        return []
        
    min_heap = []
    # 步骤 1：往空堆中不断插入元素建立最小堆
    for num in arr:
        heapq.heappush(min_heap, num)
        
    # 步骤 2：调用 k 次 Extract-min 按顺序输出最小的 k 个数
    result = []
    for _ in range(min(k, len(arr))): 
        result.append(heapq.heappop(min_heap))
        
    return result

# ================= 测试 =================
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
result = get_k_smallest_by_heap(arr, k)

arr_str = "[" + ",".join(map(str, arr)) + "]"
result_str = ", ".join(map(str, result))
print(f"数组arr={arr_str} 的前{k}个最小数据为：{result_str}")
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

### 问题 6 答案

**算法设计思路**

使用**栈 (Stack)** 来模拟星球的存续情况。
* 遍历星球数组 `A`，如果栈为空，或当前星球向右运动（`>0`），或栈顶星球向左运动（`<0`），说明不会发生碰撞，直接将星球压入栈。
* 发生碰撞的条件：栈顶 `> 0` 且 当前星球 `< 0`。
    * 栈顶质量 < 当前质量：栈顶被摧毁弹出，继续用当前星球与新的栈顶判断（使用 `while` 循环）。
    * 栈顶质量 == 当前质量：当前星球消失，栈顶星球保留。停止判断。
    * 栈顶质量 > 当前质量：当前星球被摧毁。停止判断。

**复杂度分析**

* **时间复杂度：** $O(n)$。
  * **操作次数推导（平摊分析 Amortized Analysis）：** 表面上看，代码中包含了一个 `for` 循环嵌套了一个 `while` 循环，似乎可能会达到 $O(n^2)$。但实际上我们需要计算总的操作次数。
  * 假设数组长度为 $n$。每个星球（元素）最多被压入栈（`push`）一次。
  * 一个星球只有在已经在栈中时，才可能被弹出（`pop`）。因此，整个算法执行期间，`pop` 操作的总次数绝对不可能超过 $n$ 次。
  * 综上，`for` 循环执行 $n$ 次，内部的 `while` 循环触发的 `pop` 操作**全局累计**最多 $n$ 次。总的基本操作次数严格限制在 $2n$ 以内，故平摊到每个元素的时间复杂度为严格的 $O(n)$。
* **空间复杂度：** $O(n)$。
  * 在最坏的情况下（例如所有星球都向右运动，或者所有向左运动的星球都在向右运动的星球的左侧），永远不会发生碰撞。此时所有的 $n$ 个星球都会被持续压入栈中，占用 $O(n)$ 的辅助空间。

````python
def planet_collisions(A):
    """
    模拟星球碰撞过程
    :param A: List[int], 星球质量和方向数组
    :return: List[int], 最终存续的星球数组
    """
    stack = []
    
    for planet in A:
        alive = True # 标记当前遍历到的星球是否存活
        
        # 发生碰撞的条件：栈不为空 且 栈顶星球向右运动(>0) 且 当前星球向左运动(<0)
        while alive and stack and stack[-1] > 0 and planet < 0:
            mass_top = stack[-1]
            mass_curr = abs(planet)
            
            if mass_top < mass_curr:
                # 栈顶星球质量小，被摧毁弹出；当前星球继续向左飞，进入下一次while循环检查
                stack.pop()
            elif mass_top == mass_curr:
                # 质量相等：按题意，自右向左(当前星球)消失，自左向右(栈顶)保留
                alive = False
            else:
                # 栈顶星球质量大，当前星球被摧毁
                alive = False
                
        # 如果当前星球历经碰撞后仍然存活，则将其压入栈中
        if alive:
            stack.append(planet)
            
    return stack

# ================= 题目中的测试示例 =================

# 示例 1
A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
output1 = planet_collisions(A1)
print(f"输入: A={A1}")
print(f"输出: {output1}\n")

# 示例 2
A2 = [23, -8, 9, -3, -7, 9, -23, 22]
output2 = planet_collisions(A2)
print(f"输入: A={A2}")
print(f"输出: {output2}")
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

### 问题 7 答案（此题求助了大模型）

**算法设计思路**

解决 0-1 背包问题最常用的算法是**动态规划 (Dynamic Programming)**。

* **状态定义：** 我们定义一个一维数组 `dp`，其中 `dp[j]` 表示当小推车承重为 `j` 时，所能装入大理石的最大总价值。
* **状态转移方程：** 遍历每一块大理石（假设当前大理石重量为 `w`，价值为 `v`），对于每一个容量 `j`（从最大承重逆序遍历到 `w`），我们都有两种选择：
    1.  **不装这块大理石：** 最大价值保持不变，即 `dp[j]`。
    2.  **装入这块大理石：** 最大价值变为“腾出这块大理石重量后的最大价值”加上“当前大理石的价值”，即 `dp[j - w] + v`。
    综合这两种选择，我们取最大值来更新当前容量下的最优解：
    $$dp[j] = \max(dp[j], dp[j - w] + v)$$
* **逆序遍历的原因：** 在内层循环中，容量 `j` 必须从最大承重 `W` 向后递减到当前物品重量 `w`。这是为了保证在更新 `dp[j]` 时，所用到的 `dp[j - w]` 是**上一轮（即还没考虑当前物品时）** 的状态，从而确保每块大理石最多只被放入一次。

**复杂度分析**

* **时间复杂度：** $O(N \times W)$。
  * **操作次数推导：** 算法由两层嵌套循环构成。外层循环遍历所有物品，执行次数为物品总数 $N$。
  * 内层循环变量 `j` 从最大承重 $W$ 递减到当前物品的重量 $w_i$。在最坏情况下（假设物品重量很小），内层循环的执行次数接近于 $W$。
  * 每次内层循环执行的都是基础的数组索引读取、加法运算和 `max()` 比较操作，耗时均为 $\Theta(1)$。
  * 因此，总的操作次数上界为 $\sum_{i=1}^{N} W = N \times W$。时间复杂度为 $O(N \times W)$。
* **空间复杂度：** $O(W)$。
  * 经典的二维 DP 状态表需要 $O(N \times W)$ 的空间。但由于我们应用了**状态压缩**，当前层的状态 $dp[j]$ 只依赖于上一层的 $dp[j]$ 和 $dp[j-w]$，因此我们可以复用一个长度为 $W + 1$ 的一维数组，将辅助空间复杂度极大地优化到了 $O(W)$。

````python
def knapsack_01(items, max_capacity):
    """
    解决 0-1 背包问题
    :param items: List[tuple], 包含 (重量, 价值) 的元组列表
    :param max_capacity: int, 背包/小推车的最大承重
    :return: int, 能装入的最大价值
    """
    # 初始化 dp 数组，长度为 max_capacity + 1，初始值均为 0
    dp = [0] * (max_capacity + 1)
    
    # 遍历每块大理石
    for weight, value in items:
        # 从最大容量开始逆序遍历，直到容量刚好能装下当前大理石
        for j in range(max_capacity, weight - 1, -1):
            # 状态转移方程：选择不装当前物品 vs 装当前物品
            dp[j] = max(dp[j], dp[j - weight] + value)
            
    # dp[max_capacity] 即为最大承重下的最大价值
    return dp[max_capacity]

# ================= 测试运行 =================
# 题目给定的数据：(重量, 价值)
marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
cart_capacity = 9

max_value = knapsack_01(marbles, cart_capacity)

print(f"大理石数据 (重量, 价值): {marbles}")
print(f"小推车最大承重: {cart_capacity}")
print(f"使用这个小推车装大理石的最大价值为: {max_value}")
````

## 问题 8

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

### 问题 8 答案（此题求助了大模型）

**算法设计思路**

可以通过一次中序遍历来找到被错误交换的两个节点。具体思路如下：

1.  **初始化变量**：设定三个指针：
    * `prev`：用于在中序遍历中记录上一个访问的节点，初始值设为极小值。
    * `first`：记录第一个错误节点。
    * `second`：记录第二个错误节点。
2.  **寻找逆序对**：在遍历过程中，不断比较 `prev.val` 和当前节点 `curr.val`。如果出现 `prev.val > curr.val`，说明找到了逆序对。
    * **第一种情况（两个错误节点相邻）**：例如中序序列原本是 `[1, 2, 3, 4]`，交换 2 和 3 后变成 `[1, 3, 2, 4]`。这里只有一个逆序对 `3 > 2`。被交换的节点就是这个逆序对中的两个节点。
    * **第二种情况（两个错误节点不相邻）**：例如中序序列原本是 `[1, 2, 3, 4, 5]`，交换 2 和 5 后变成 `[1, 5, 3, 4, 2]`。这里会出现两个逆序对：`5 > 3` 和 `4 > 2`。被交换的节点分别是**第一个逆序对的前一个节点（5）**和**第二个逆序对的后一个节点（2）**。
3.  **记录节点**：
    * 当第一次遇到 `prev.val > curr.val` 时，我们将 `prev` 赋给 `first`，将 `curr` 赋给 `second`。
    * 当第二次遇到 `prev.val > curr.val` 时，我们只需更新 `second` 为当前的 `curr`。
4.  **恢复树**：遍历结束后，直接交换 `first` 和 `second` 的值即可完成树的恢复，无需改变树的拓扑结构。

**复杂度分析**

* **时间复杂度：** $O(n)$。
  * **操作次数推导：** 该算法的核心是一个标准的中序遍历（In-order Traversal）。对于一棵包含 $n$ 个节点的二叉树，中序遍历会严格按照左子树、根节点、右子树的顺序，访问且仅访问每个节点恰好一次。
  * 在访问每个节点时，执行的操作仅为常数级别的 $\Theta(1)$ 操作（包括比较当前节点与 `prev` 节点的值，以及最多两次的指针赋值更新）。
  * 设树节点数为 $n$，则总执行时间 $T(n) = n \times \Theta(1) = \Theta(n)$，即时间复杂度为 $O(n)$。
* **空间复杂度：** $O(h)$，其中 $h$ 为树的高度。
  * 该算法没有开辟额外的数组或哈希表，空间开销全部来自于递归函数 `inorder` 的隐式系统调用栈。
  * 递归栈的最大深度等于二叉树的高度 $h$。
  * **最坏情况：** 树退化为一条链表（所有节点只有左子树或只有右子树），此时高度 $h = n$，空间复杂度为 $O(n)$。
  * **最好情况：** 树是完全平衡的二叉搜索树，此时高度 $h = \log_2 n$，空间复杂度为 $O(\log n)$。

````python
import json

# 1. 定义树节点结构
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# 2. 核心算法类
class Solution:
    def recoverTree(self, root: TreeNode) -> None:
        self.first = None
        self.second = None
        self.prev = TreeNode(float('-inf')) 
        
        def inorder(node):
            if not node:
                return
            
            inorder(node.left)
            
            # 检查是否满足递增性质
            if self.prev.val > node.val:
                # 记录第一个错误节点
                if not self.first:
                    self.first = self.prev
                # 记录（或更新）第二个错误节点
                self.second = node
            
            self.prev = node
            inorder(node.right)
            
        inorder(root)
        
        # 交换两个错误节点的值
        if self.first and self.second:
            self.first.val, self.second.val = self.second.val, self.first.val

# 3. 辅助函数：将二叉树转为层序遍历列表
def tree_to_list(root):
    if not root: return []
    res, queue = [], [root]
    while queue:
        node = queue.pop(0)
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None: res.pop() # 去掉末尾多余的空节点
    return res

# 4. 辅助函数：完美打印 null
def print_tree(root):
    # 使用 json.dumps 可以将 Python 的 None 完美转换为小写的 null
    return json.dumps(tree_to_list(root)).replace(" ", "")

# ================= 自动运行测试用例部分 =================

if __name__ == "__main__":
    sol = Solution()

    # --- 测试用例 1 ---
    # 输入: [1,3,null,null,2] (3和1被交换了)
    root1 = TreeNode(1)
    root1.left = TreeNode(3)
    root1.left.right = TreeNode(2)
    
    print("测试用例 1:")
    print(f"恢复前: {print_tree(root1)}")
    sol.recoverTree(root1)
    print(f"恢复后: {print_tree(root1)}")
    print("-" * 30)

    # --- 测试用例 2 ---
    # 输入: [3,1,4,null,null,2] (3和2被交换了)
    root2 = TreeNode(3)
    root2.left = TreeNode(1)
    root2.right = TreeNode(4)
    root2.right.left = TreeNode(2)
    
    print("测试用例 2:")
    print(f"恢复前: {print_tree(root2)}")
    sol.recoverTree(root2)
    print(f"恢复后: {print_tree(root2)}")

# your algorithm time complexity is: O(n)
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

### 问题 10 答案（此题求助了大模型）

**算法设计思路**

此题可以通过**深度优先搜索 (DFS)** 解决。
从根节点开始遍历树。每当我们向下走一层，当前路径代表的数字就会由于进位而乘以 10，然后加上当前节点的值。
1. **状态传递**：给递归函数传入当前节点 `node` 和到该节点为止计算出的数字 `current_sum`。进入节点时更新：`current_sum = current_sum * 10 + node.val`。
2. **终止条件**：如果当前节点是叶子节点（即没有左右子树），那么就将当前的 `current_sum` 向上返回。
3. **递归计算**：如果不是叶子节点，就将左子树和右子树生成的数字之和返回相加。

**复杂度分析**

* **时间复杂度：** $O(n)$。
  * **操作次数推导：** 算法采用了深度优先搜索（DFS）来遍历完整棵二叉树。无论树的形态如何，DFS 必须访问每一个存在的节点恰好一次，以确保不会漏掉任何一条路径。
  * 对于每一个被访问的节点，算法执行的操作是固定的：一次乘法（`current_sum * 10`）、一次加法（`+ node.val`）、以及判断是否为叶子节点的布尔逻辑运算。这些都是 $\Theta(1)$ 的基本运算。
  * 因此，对于一棵含有 $n$ 个节点的树，总计算量严格等于 $n \times \Theta(1)$，时间复杂度为 $O(n)$。
* **空间复杂度：** $O(h)$，其中 $h$ 为树的高度。
  * 虽然我们在函数参数中传递了 `current_sum`，但这只是按值传递的标量变量，不占用动态结构空间。真正的空间消耗来自于 DFS 递归产生的系统调用栈。
  * 递归栈的深度即为当前节点所在的层数，其最大深度等于树的高度 $h$。
  * **最坏情况：** 树呈链状（例如所有节点都在左边），高度 $h = n$，空间复杂度为 $O(n)$。
  * **最好/平均情况：** 树是分布均匀的二叉树，高度 $h = \log_2 n$，空间复杂度降为 $O(\log n)$。

````python
# 定义树节点结构
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# 核心算法类
class Solution:
    def sumNumbers(self, root: TreeNode) -> int:
        def dfs(node, current_sum):
            if not node:
                return 0
                
            # 计算到达当前节点时的组合数字
            current_sum = current_sum * 10 + node.val
            
            # 如果是叶子节点，直接返回当前形成的完整数字
            if not node.left and not node.right:
                return current_sum
                
            # 不是叶子节点，则递归求解左右子树之和
            return dfs(node.left, current_sum) + dfs(node.right, current_sum)
            
        return dfs(root, 0)

# ================= 测试运行部分 =================

if __name__ == "__main__":
    sol = Solution()

    # 测试用例 1：root = [1,2,3]
    # 构建过程：
    #     1
    #    / \
    #   2   3
    root1 = TreeNode(1)
    root1.left = TreeNode(2)
    root1.right = TreeNode(3)
    
    result1 = sol.sumNumbers(root1)
    print(f"输入: root = [1,2,3]")
    print(f"输出: {result1}")
    print("-" * 30)

    # 测试用例 2：root = [4,9,0,5,1]
    # 构建过程：
    #       4
    #      / \
    #     9   0
    #    / \
    #   5   1
    root2 = TreeNode(4)
    root2.left = TreeNode(9)
    root2.right = TreeNode(0)
    root2.left.left = TreeNode(5)
    root2.left.right = TreeNode(1)
    
    result2 = sol.sumNumbers(root2)
    print(f"输入: root = [4,9,0,5,1]")
    print(f"输出: {result2}")
    
# your algorithm time complexity is: O(n)
````
